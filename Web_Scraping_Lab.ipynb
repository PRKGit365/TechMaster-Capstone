## Extract information from the given web site

You will extract the data from the below web site: <br>


```python
#this url contains the data you need to scrape
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
```

The data you need to scrape is the **name of the programming language** and **average annual salary**.<br> It is a good idea to open the url in your web broswer and study the contents of the web page before you start to scrape.


Import the required libraries


```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

Download the webpage at the url


```python
data = requests.get("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html").text
```

Create a soup object


```python
soup = BeautifulSoup(data, 'html.parser')
```

Scrape the `Language name`, `Created By` `annual average salary`and `Learning Difficulty`.


```python
#your code goes here
from bs4 import BeautifulSoup
import requests
import pandas as pd

table = soup.find('table')
data_list = []

# Loop through each row in the table
for row in table.find_all('tr'):
    cols = row.find_all('td')
    # Skip the header row if necessary
    if cols[1].getText() == 'Language':
        continue

    Language_name = cols[1].getText()
    Created_by = cols[2].getText()
    Annual_average_salary = cols[3].getText()
    Learning_Difficulty = cols[4].getText()

    print('Language name= {}, Created By = {}, Annual average salary= {} and Learning Difficulty = {}'.format(Language_name, Created_by, Annual_average_salary, Learning_Difficulty))

    data_list.append({
        'Language name': Language_name,
        'Created By': Created_by,
        'annual average salary': Annual_average_salary,
        'Learning Difficulty': Learning_Difficulty
    })

table_df = pd.DataFrame(data_list, columns=['Language name', 'Created By', 'annual average salary', 'Learning Difficulty'])
table_df
```

Create a _dataframe_ for scrapped data


```python
import pandas as pd
# We can define df using table_df from the scraping step
try:
    df = table_df.copy()
except NameError:
    df = pd.DataFrame(data_list)
display(df)
```

Save the scrapped data into a file named _popular-languages.csv_


```python
df.to_csv('popular-languages.csv', index=False)
print("Saved successfully to popular-languages.csv")
```
